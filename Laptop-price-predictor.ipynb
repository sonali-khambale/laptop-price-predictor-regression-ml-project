```python
# Convert Memory column to string and remove '.0'
df['Memory'] = df['Memory'].astype(str).replace(r'\.0', '', regex=True)

# Remove GB and convert TB into GB format
df['Memory'] = df['Memory'].str.replace('GB', '')
df['Memory'] = df['Memory'].str.replace('TB', '000')

# Split storage into two parts using '+'
new = df['Memory'].str.split('+', n=1, expand=True)

# First storage part
df['first'] = new[0].str.strip()

# Second storage part
df['second'] = new[1]

# Identify storage type in first part
df['Layer1HDD'] = df['first'].apply(lambda x: 1 if 'HDD' in x else 0)
df['Layer1SSD'] = df['first'].apply(lambda x: 1 if 'SSD' in x else 0)
df['Layer1Hybrid'] = df['first'].apply(lambda x: 1 if 'Hybrid' in x else 0)
df['Layer1Flash_Storage'] = df['first'].apply(lambda x: 1 if 'Flash Storage' in x else 0)

# Keep only numbers in first part
df['first'] = df['first'].str.replace(r'\D', '', regex=True)

# Fill missing second part with 0
df['second'] = df['second'].fillna('0')

# Identify storage type in second part
df['Layer2HDD'] = df['second'].apply(lambda x: 1 if 'HDD' in x else 0)
df['Layer2SSD'] = df['second'].apply(lambda x: 1 if 'SSD' in x else 0)
df['Layer2Hybrid'] = df['second'].apply(lambda x: 1 if 'Hybrid' in x else 0)
df['Layer2Flash_Storage'] = df['second'].apply(lambda x: 1 if 'Flash Storage' in x else 0)

# Keep only numbers in second part
df['second'] = df['second'].str.replace(r'\D', '', regex=True)

# Convert to integer
df['first'] = df['first'].astype(int)
df['second'] = df['second'].astype(int)

# Create final storage columns
df['HDD'] = (df['first'] * df['Layer1HDD'] +
              df['second'] * df['Layer2HDD'])

df['SSD'] = (df['first'] * df['Layer1SSD'] +
             df['second'] * df['Layer2SSD'])

df['Hybrid'] = (df['first'] * df['Layer1Hybrid'] +
                df['second'] * df['Layer2Hybrid'])

df['Flash_Storage'] = (df['first'] * df['Layer1Flash_Storage'] +
                       df['second'] * df['Layer2Flash_Storage'])

# Drop unnecessary columns
df.drop(columns=[
    'first', 'second',
    'Layer1HDD', 'Layer1SSD', 'Layer1Hybrid', 'Layer1Flash_Storage',
    'Layer2HDD', 'Layer2SSD', 'Layer2Hybrid', 'Layer2Flash_Storage'
], inplace=True)
```

# Train Test Split

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

# *modelling

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import r2_score,mean_absolute_error
```

```python
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor,GradientBoostingRegressor,AdaBoostRegressor,ExtraTreesRegressor
from sklearn.svm import SVR
```

### Linear Regression

```python
# Column Transformer
step1 = ColumnTransformer(transformers=[
    ('col_tnf', OneHotEncoder(sparse_output=False, drop='first'),
     [0,1,7,10,11])
], remainder='passthrough')

# Model
step2 = LinearRegression()

# Pipeline
pipe = Pipeline([
    ('step1', step1),
    ('step2', step2)
])

# Train model
pipe.fit(X_train, y_train)

# Prediction
y_pred = pipe.predict(X_test)

# Evaluation
print('R2 Score:', r2_score(y_test, y_pred))
print('MAE:', mean_absolute_error(y_test, y_pred))
```

### Voting Regressor

```python
from sklearn.ensemble import VotingRegressor, RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor, ExtraTreesRegressor

# Column Transformer
step1 = ColumnTransformer(transformers=[
    ('col_tnf',
     OneHotEncoder(sparse_output=False, drop='first'),
     [0,1,7,10,11])
], remainder='passthrough')

# Models
rf = RandomForestRegressor(
    n_estimators=350,
    random_state=3,
    max_samples=0.5,
    max_features=0.75,
    max_depth=15,
    bootstrap=True
)

gbdt = GradientBoostingRegressor(
    n_estimators=100,
    max_features=0.5
)

et = ExtraTreesRegressor(
    n_estimators=100,
    random_state=3,
    max_samples=0.5,
    max_features=0.75,
    max_depth=10,
    bootstrap=True
)

# Voting Regressor
step2 = VotingRegressor(
    estimators=[
        ('rf', rf),
        ('gbdt', gbdt),
        ('et', et)
    ],
    weights=[5,1,1]
)

# Pipeline
pipe = Pipeline([
    ('step1', step1),
    ('step2', step2)
])

# Train model
pipe.fit(X_train, y_train)

# Prediction
y_pred = pipe.predict(X_test)

# Evaluation
print("R2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
```

### Stacking

```python
from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import Ridge

# Base models
estimators = [
    ('rf', RandomForestRegressor(
        n_estimators=350,
        random_state=3,
        max_samples=0.5,
        max_features=0.75,
        max_depth=15,
        bootstrap=True
    )),
    
    ('gbdt', GradientBoostingRegressor(
        n_estimators=100,
        max_features=0.5
    )),
    
    ('et', ExtraTreesRegressor(
        n_estimators=100,
        random_state=3,
        max_samples=0.5,
        max_features=0.75,
        max_depth=10,
        bootstrap=True
    ))
]

# Column Transformer
step1 = ColumnTransformer(transformers=[
    ('col_tnf',
     OneHotEncoder(sparse_output=False, drop='first'),
     [0,1,7,10,11])
], remainder='passthrough')

# Stacking Regressor
step2 = StackingRegressor(
    estimators=estimators,
    final_estimator=Ridge(alpha=100)
)

# Pipeline
pipe = Pipeline([
    ('step1', step1),
    ('step2', step2)
])

# Train model
pipe.fit(X_train, y_train)

# Prediction
y_pred = pipe.predict(X_test)

# Evaluation
print("R2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
```
